```python
import numpy as np

# 1. Membuat Array dari List biasa
harga_barang = np.array([50000, 120000, 75000, 200000, 90000])

# 2. Vectorization (Hitung sekaligus tanpa 'for' loop)
# Misal semua barang kena diskon 10% (dikali 0.9)
harga_diskon = harga_barang * 0.9
print("Harga setelah diskon:", harga_diskon)

# 3. Statistik Dasar (Wajib dikuasai Data Analyst)
print("Total semua barang :", np.sum(harga_barang))
print("Rata-rata (Mean)   :", np.mean(harga_barang))
print("Nilai Tengah (Med) :", np.median(harga_barang))
print("Termahal (Max)     :", np.max(harga_barang))

# 4. Filtering Array (Cari barang di atas 80 ribu)
barang_mahal = harga_barang[harga_barang > 80000]
print("Barang > 80rb      :", barang_mahal)
```
