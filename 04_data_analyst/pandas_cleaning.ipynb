```python
import pandas as pd

df = pd.read_csv("data/sales.csv")
print("Jumlah baris awal:", len(df))

# 1. Hapus baris duplikat berdasarkan id_transaksi
df = df.drop_duplicates(subset=["id_transaksi"])
print("Setelah hapus duplikat:", len(df))

# 2. Mengisi data kosong (Missing Values / NaN)
df["kategori"] = df["kategori"].fillna("Lainnya")
df["metode_pembayaran"] = df["metode_pembayaran"].fillna("Tidak Diketahui")

# 3. Merapikan teks kotor (hapus spasi berlebih & samakan huruf besar)
# Contoh: "qris " menjadi "QRIS"
df["metode_pembayaran"] = df["metode_pembayaran"].str.strip().str.upper()

# 4. Membersihkan format mata uang teks menjadi angka bulat (Integer)
# Contoh: "Rp 250.000" -> "250000" -> 250000
df["harga_satuan"] = (
    df["harga_satuan"]
    .astype(str)
    .str.replace("Rp", "", regex=False)
    .str.replace(".", "", regex=False)
    .str.strip()
    .astype(int)
)

# 5. Menyeragamkan format tanggal yang campur-aduk menjadi tipe Datetime
df["tanggal"] = pd.to_datetime(df["tanggal"], format="mixed")

print("\n--- Hasil Data Setelah Dibersihkan ---")
print(df[["tanggal", "harga_satuan", "kategori", "metode_pembayaran"]].head())
```
