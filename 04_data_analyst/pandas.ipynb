```python
import pandas as pd

# 1. Membaca file CSV menjadi DataFrame
df = pd.read_csv("data/sales.csv")

# 2. Mengintip isi data (WAJIB dilakukan pertama kali)
print("--- 5 Baris Pertama ---")
print(df.head())

print("\n--- Info Tipe Data & Jumlah Baris ---")
print(df.info())

# 3. Memilih Kolom Tertentu
subset = df[["tanggal", "nama_produk", "qty", "total_harga"]]

# 4. Filtering Data (Seperti WHERE di SQL)
# Cari transaksi yang statusnya 'Selesai' DAN qty lebih dari 2
transaksi_besar = df[(df["status_pengiriman"] == "Selesai") & (df["qty"] > 2)]
print(f"\nJumlah transaksi selesai dengan qty > 2: {len(transaksi_besar)}")

# 5. Grouping & Agregasi (Seperti GROUP BY di SQL)
# Hitung total pendapatan per nama produk
omzet_produk = df.groupby("nama_produk")["total_harga"].sum().sort_values(ascending=False)
print("\n--- Total Omzet per Produk ---")
print(omzet_produk)
```
