```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

os.makedirs("reports", exist_ok=True)
df = pd.read_csv("data/sales.csv").drop_duplicates(subset=["id_transaksi"])
df["tanggal"] = pd.to_datetime(df["tanggal"], format="mixed")

# Filter hanya transaksi yang berhasil selesai
df_selesai = df[df["status_pengiriman"] == "Selesai"].copy()

# 1. Grafik Batang (Bar Chart): Membandingkan Omzet antar Produk
omzet = df_selesai.groupby("nama_produk")["total_harga"].sum().reset_index()
omzet = omzet.sort_values("total_harga", ascending=False)

plt.figure(figsize=(10, 5))
sns.barplot(data=omzet, x="total_harga", y="nama_produk", palette="viridis")
plt.title("Total Pendapatan per Produk (Transaksi Selesai)", fontweight="bold")
plt.xlabel("Total Pendapatan (Rupiah)")
plt.ylabel("Nama Produk")
plt.tight_layout()
plt.savefig("reports/grafik_produk.png", dpi=150)
plt.close()

# 2. Grafik Garis (Line Chart): Melihat Tren Pendapatan per Bulan
df_selesai["bulan"] = df_selesai["tanggal"].dt.to_period("M").astype(str)
tren_bulanan = df_selesai.groupby("bulan")["total_harga"].sum().reset_index()

plt.figure(figsize=(10, 4))
sns.lineplot(data=tren_bulanan, x="bulan", y="total_harga", marker="o", color="teal")
plt.title("Tren Pendapatan Bulanan Tahun 2026", fontweight="bold")
plt.xlabel("Bulan")
plt.ylabel("Pendapatan (Rupiah)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig("reports/grafik_tren.png", dpi=150)
plt.close()

print("Grafik berhasil disimpan di folder reports/!")
```
