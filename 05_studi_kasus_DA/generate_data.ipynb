```python
import os
import random
from datetime import datetime, timedelta
import numpy as np
import pandas as pd

os.makedirs("data", exist_ok=True)
np.random.seed(99)
random.seed(99)

# 1. Generate menu_kopi.csv
menu_data = [
    ("MN-01", "Spanish Latte", "Coffee", 28000, 11000),
    ("MN-02", "Palm Sugar Latte", "Coffee", 25000, 9500),
    ("MN-03", "Caramel Macchiato", "Coffee", 32000, 13000),
    ("MN-04", "Americano", "Coffee", 20000, 6000),
    ("MN-05", "Matcha Latte", "Non-Coffee", 30000, 12500),
    ("MN-06", "Chocolate Hazelnut", "Non-Coffee", 29000, 11500),
    ("MN-07", "Butter Croissant", "Pastry", 24000, 10000),
    ("MN-08", "Almond Pain au Chocolat", "Pastry", 28000, 12000),
]
df_menu = pd.DataFrame(menu_data, columns=["id_menu", "nama_menu", "kategori", "harga_jual", "modal"])
df_menu.to_csv("data/menu_kopi.csv", index=False)

# 2. Generate pesanan_kopi.csv (Kotor)
cabang_list = ["Jakarta Timur", "Jakarta Selatan", "Bekasi", "jakarta timur ", " BEKASI", np.nan]
tipe_order = ["Dine-In", "Takeaway", "GoFood", "GrabFood", "gofood "]
status_list = ["Selesai", "Selesai", "Selesai", "Selesai", "Batal"]

pesanan = []
for i in range(1, 1201):
    menu = random.choice(menu_data)
    waktu = datetime(2026, 6, 1) + timedelta(
        days=random.randint(0, 150),
        hours=random.randint(7, 22),
        minutes=random.randint(0, 59)
    )
    qty = random.randint(1, 4)
    if random.random() < 0.015:
        qty = random.randint(15, 30)  # Outlier pesanan kantor/acara

    pesanan.append({
        "id_order": f"ORD-{i:04d}",
        "waktu_transaksi": waktu.strftime("%Y-%m-%d %H:%M") if random.random() > 0.2 else waktu.strftime("%d/%m/%Y %H:%M"),
        "id_menu": menu[0],
        "qty": qty,
        "cabang": random.choice(cabang_list),
        "tipe_order": random.choice(tipe_order) if random.random() > 0.03 else np.nan,
        "status": random.choice(status_list)
    })

df_pesanan = pd.DataFrame(pesanan)
# Tambah 25 duplikat
df_pesanan = pd.concat([df_pesanan, df_pesanan.sample(25, random_state=99)], ignore_index=True)
df_pesanan.to_csv("data/pesanan_kopi.csv", index=False)

print("Dataset Kedai Kopi siap! File tersimpan di data/menu_kopi.csv & data/pesanan_kopi.csv")
```
